```python
import matplotlib.pyplot as plt

# Extract year for trend analysis
df['Year'] = df['Release_Date'].dt.year

# 1. Content Type Trend Analysis
plt.figure(figsize=(12,6))
content_trend = df.groupby(['Year','Type']).size().unstack(fill_value=0)

# Plot the trend lines
content_trend.plot(kind='line', marker='o', figsize=(12,6))

# Title and labels
plt.title("Netflix Content Type Trend Over Years", fontsize=14)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Number of Titles", fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)

# ✅ Move legend outside the plot
plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc='upper left',
    borderaxespad=0.,
    title="Type",
    fontsize=8
)

# Adjust layout to make space for legend
plt.tight_layout(rect=[0, 0, 0.8, 1])

# Show plot
plt.show()
```

```python
# ---------------------------------
# 2. Genre Popularity and Evolution
# ---------------------------------
# Assuming 'Category' is the genre
top_genres = df['Category'].value_counts().nlargest(10).index  # Top 10 genres
genre_trend = df[df['Category'].isin(top_genres)].groupby(['Year','Category']).size().unstack(fill_value=0)

plt.figure(figsize=(14,7))
genre_trend.plot(marker='o')
plt.title("Top 10 Genre Popularity Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.grid(True)
plt.legend(title='Genre', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# ---------------------------------
# 3. Country-Wise Content Contribution
# ---------------------------------
# Top 10 countries by content count
top_countries = df['Country'].value_counts().nlargest(10).index
country_counts = df[df['Country'].isin(top_countries)]['Country'].value_counts()

plt.figure(figsize=(10,6))
sns.barplot(x=country_counts.values, y=country_counts.index, palette='viridis')
plt.title("Top 10 Countries by Netflix Content Contribution")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python

# -------------------------------
# Top Directors
top_directors = df['Director'].value_counts().nlargest(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_directors.values, y=top_directors.index, palette='magma')
plt.title("Top 10 Directors by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.show()

# Top Cast Members
df_cast = df.copy()
df_cast['Cast'] = df_cast['Cast'].fillna('Unknown')
df_cast = df_cast.assign(Cast=df_cast['Cast'].str.split(',')).explode('Cast')
df_cast['Cast'] = df_cast['Cast'].str.strip()

top_cast = df_cast['Cast'].value_counts().nlargest(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_cast.values, y=top_cast.index, palette='coolwarm')
plt.title("Top 10 Cast Members by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Cast Member")
plt.show()
```
