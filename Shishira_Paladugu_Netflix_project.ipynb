```python
import matplotlib.pyplot as plt

df['Year'] = df['Release_Date'].dt.year

# 1. Content Type Trend Analysis - Focusing on top categories
# Get top 10 categories
top_categories = df['Type'].value_counts().nlargest(10).index

# Filter the DataFrame to include only the top categories
df_top_categories = df[df['Type'].isin(top_categories)].copy()

content_trend = df_top_categories.groupby(['Year', 'Type']).size().unstack(fill_value=0)

# Create figure with constrained layout
fig, ax = plt.subplots(figsize=(15, 8), constrained_layout=True)

# Plot the trend lines
content_trend.plot(kind='line', marker='o', ax=ax)

# Title and labels
ax.set_title("Netflix Content Type Trend Over Years (Top 10 Categories)", fontsize=14)
ax.set_xlabel("Year", fontsize=12)
ax.set_ylabel("Number of Titles", fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.yticks(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

# Move legend outside the plot
ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    borderaxespad=0,
    title="Type",
    fontsize=10
)

# Display the plot
plt.show()
```

```python
# ---------------------------------
# 2. Genre Popularity and Evolution
# ---------------------------------
# Assuming 'Category' is the genre
top_genres = df['Category'].value_counts().nlargest(10).index  # Top 10 genres
genre_trend = df[df['Category'].isin(top_genres)].groupby(['Year','Category']).size().unstack(fill_value=0)

plt.figure(figsize=(14,7))
genre_trend.plot(marker='o')
plt.title("Top 10 Genre Popularity Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.grid(True)
plt.legend(title='Genre', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# ---------------------------------
# 3. Country-Wise Content Contribution
# ---------------------------------
# Top 10 countries by content count
top_countries = df['Country'].value_counts().nlargest(10).index
country_counts = df[df['Country'].isin(top_countries)]['Country'].value_counts()

plt.figure(figsize=(10,6))
sns.barplot(x=country_counts.values, y=country_counts.index, palette='viridis', hue=country_counts.index, legend=False)
plt.title("Top 10 Countries by Netflix Content Contribution")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
# -------------------------------
# Top Directors
top_directors = df['Director'].value_counts().nlargest(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_directors.values, y=top_directors.index, palette='magma', hue=top_directors.index, legend=False)
plt.title("Top 10 Directors by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Director")
plt.show()

# Top Cast Members
df_cast = df.copy()
df_cast['Cast'] = df_cast['Cast'].fillna('Unknown')
df_cast = df_cast.assign(Cast=df_cast['Cast'].str.split(',')).explode('Cast')
df_cast['Cast'] = df_cast['Cast'].str.strip()

top_cast = df_cast['Cast'].value_counts().nlargest(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_cast.values, y=top_cast.index, palette='coolwarm', hue=top_cast.index, legend=False)
plt.title("Top 10 Cast Members by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Cast Member")
plt.show()
```
